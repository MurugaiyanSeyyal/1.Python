```python
# print 0 to 20 by using range

for num in range(0,20):
    print(num)
```

```python
# print range 10 to 20

for num in range(10,20):
    print(num, end=" ")
```

```python
# Print number of items in the list by using len

lists = [10, 20, 14, 55, 43, 87, 76]
print(lists)
print("Number of items in the Lists:",len(lists))
```

```python
# Artificial Intelligence

text = "Artificial Intelligence"
print(text)
for char in text:
    print(char)
```

```python
# Loop

labels = ["-Your Name-","-Your Age-","-Your Profession-"]
for label in labels:
    print(label)
```

```python
# Print this mixered datatype using Tuples

my_tuple = (1, 'Welcome', 2, 'Hope')
print(my_tuple)
```

```python
my_tuple = (0, 1, 2, 3),('Python','Hope')
print(my_tuple)
```

```python
# Print Odd Numbers in the list

numbers = [20,10,16,19,25,1,276,188]
print(numbers)
for num in numbers:
    if(num % 2 !=0):
        print(num,"is odd")
```

```python
# Print Even Numbers in the list

numbers = [20,10,16,19,25,1,276,188]
print(numbers)
for num in numbers:
    if(num % 2 ==0):
        print(num,"is Even")
```
