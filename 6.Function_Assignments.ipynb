```python
# Create a Funtion to find Odd or Even Number:

def OddEven():
    num = int(input("Enter the Number: "))
    if ((num % 2) ==0):
        print(num, "is Even Number")
    else:
        print(num, "is Odd Number")
        
```

```python
# Create a Function to find Marriage Eligible

def Elegible(gender,age):
    if (gender == "Male") and (age >= 21):
        return "Elegible"
    elif (gender == "Female") and (age >= 18):
        return "Elegible"
    else:
        return "Not Elegible"
```

```python
gender = input("Your Gender: ")
age = int(input("Your Age: "))
print(Elegible(gender,age))
```

```python
# Create Function to find Persentage:

def percentage(s1,s2,s3,s4,s5):
    total = s1 + s2 + s3 + s4 + s5
    per = total / 5
    return total, per
```

```python
Subjects1 = int(input("Subjects-1 = "))
Subjects2 = int(input("Subjects-2 = "))
Subjects3 = int(input("Subjects-3 = "))
Subjects4 = int(input("Subjects-4 = "))
Subjects5 = int(input("Subjects-5 = "))

total, per = percentage(Subjects1, Subjects2, Subjects3, Subjects4, Subjects5)
print("Total :",total)
print("Percentage :",per)
```
